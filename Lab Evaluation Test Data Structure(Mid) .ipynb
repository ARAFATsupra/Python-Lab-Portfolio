```python
N = []
n = int(input("Enter the number of students in the first batch: "))
for i in range(n):
  element = int(input("Enter the student roll number: "))
  N.append(element)
print("The student Ids of first batch is: ",N)
M = []
m = int(input("Enter the number of students in the second batch: "))
for i in range(m):
  element = int(input("Enter the student roll number: "))
  M.append(element)
print("The student Ids of second batch is: ",M)
S = N + M
print("The merged list is: ",S)
size = len(S)
for i in range(size//2):
  S[i],S[size-i-1] = S[size-i-1],S[i]
print("The Reversed List Is: ",S)
```

```python
class Node:
  def __init__(self, data = None):
    self.data = data
    self.next = None
n = int(input("Enter the number of employees: "))
head = None
print("Enter the employee Ids = ")
for i in range(n):
  data = int(input())
  if head is None:
    head = Node(data)
  else:
    temp = head
    while temp.next is not None:
      temp = temp.next
    temp.next = Node(data)
print("The Employee Ids are: ")
temp = head
while temp is not None:
  print(temp.data, end = " ")
  temp = temp.next
print()
```
