# University Management System
You need to create a simple University Management System with classes to represent Person,
Student, and Faculty. The Person class will serve as the base class, and both Student and
Faculty will inherit from it.
Requirements:
● The Person class should have attributes for name, age, and gender, and a method
display_info() to print these details.
● The Student class should have additional attributes like student_id, major, and
year. It should have a method get_study_level() which prints whether the student
is in Undergraduate, Postgraduate, etc.
● The Faculty class should have additional attributes like faculty_id, department,
and title. It should have a method get_title() that returns the academic title (e.g.,
Professor, Assistant Professor).
● Use constructor chaining to ensure the base class constructor is called when creating
Student and Faculty objects.


```python
class Person:
  def __init__(self,name,age,gender):
    self.name = name
    self.age = age
    self.gender = gender
  def display_info(self):
    print(f"Name: {self.name}, Age: {self.age}, Gender: {self.gender}")
class Student(Person):
  def __init__(self,name,age,gender,student_id,major,year):
    super().__init__(name,age,gender)
    self.student_id = student_id
    self.major = major
    self.year = year
  def get_study_level(self):
    if self.year <= 4:
      print("Undergraduate")
    else:
      print("Postgraduate")
class Faculty(Person):
  def __init__(self,name,age,gender,faculty_id,department,title):
    super().__init__(name,age,gender)
    self.faculty_id = faculty_id
    self.department = department
    self.title = title
  def get_title(self):
    return self.title
#Object Create
Student1 = Student("Arafat",20,"Male","008","ITM",4)
Student1.display_info()
Student1.get_study_level()
Faculty1 = Faculty("Sakib",25,"Male","080","ITM","Professor")
Faculty1.display_info()
Faculty1.get_title()
```

# Problem: Smart Device Management System
You are tasked with creating a system to manage smart devices. You will create the following
classes:
● Device: A base class that represents general devices.
● WirelessDevice: A base class that represents devices with wireless capabilities.
● SmartDevice: A class that inherits from both Device and WirelessDevice to
represent a device that is both a general device and a wireless device.
Requirements:
● The Device class should have attributes like device_id and device_name. It should
have a method device_info() to display the device information.
● The WirelessDevice class should have attributes like connectivity_type (e.g.,
Wi-Fi, Bluetooth) and frequency (in GHz). It should have a method
wireless_info() to display the wireless properties.
● The SmartDevice class should inherit from both Device and WirelessDevice. It
should have an additional attribute os_type (e.g., Android, iOS) and a method
smart_info() to display all the device’s properties.
● Implement constructor chaining using super() and manually call the constructors of
both parent classes.


```python
class Device:
    def __init__(self, device_id, device_name):
        self.device_id = device_id
        self.device_name = device_name
    def device_info(self):
        return f"Device ID: {self.device_id}, Device Name: {self.device_name}"
class WirelessDevice:
    def __init__(self, connectivity_type, frequency):
        self.connectivity_type = connectivity_type
        self.frequency = frequency
    def wireless_info(self):
        return f"Connectivity Type: {self.connectivity_type}, Frequency: {self.frequency} GHz"
class SmartDevice(Device, WirelessDevice):
    def __init__(self, device_id, device_name, connectivity_type, frequency, os_type):
        Device.__init__(self, device_id, device_name)
        WirelessDevice.__init__(self, connectivity_type, frequency)
        self.os_type = os_type
    def smart_info(self):
        device_info = self.device_info()
        wireless_info = self.wireless_info()
        return f"{device_info}, {wireless_info}, OS Type: {self.os_type}"
#Object Create
D1 = Device("001", "Phone")
print(D1.device_info())
W1 = WirelessDevice("WiFi", 2.4)
print(W1.wireless_info())
S1 = SmartDevice("001", "Phone", "WiFi", 2.4, "Android")
print(S1.smart_info())
```

```python
import datetime
class Vehicle:
  def __init__(self, vehicle_id,Manufacturing_date,Number_of_wheels):
    self.vehicle_id = vehicle_id
    self.Manufacturing_date = Manufacturing_date
    self.Number_of_wheels = Number_of_wheels
  def Calculate_age(self):
    current_date = datetime.date.today()
    age = current_date.year - self.Manufacturing_date.year
    return age
class Bus(Vehicle):
  def __init__(self,vehicle_id,Manufacturing_date,Number_of_wheels,Number_of_seats,route_name):
    super().__init__(vehicle_id,Manufacturing_date,Number_of_wheels)
    self.Number_of_seats = Number_of_seats
    self.route_name = route_name
  def Show_name_route(self):
    print(f"Route Name: {self.route_name}")
class Customers(Vehicle):
  Product_list = []
  def __init__(self,vehicle_id,Manufacturing_date,Number_of_wheels,Customer_name,Customer_Id, Address, Product_list):
    super().__init__(vehicle_id,Manufacturing_date,Number_of_wheels)
    self.Customer_name = Customer_name
    self.Customer_Id = Customer_Id
    self.Address = Address
    self.Product_list = Product_list
  def Buy_product(self,Vehicle_Id):
    self.Product_list.append(Vehicle_Id)
    print(f"Product List: {self.Product_list}")
  def Is_premium(self):
    if len(self.Product_list) > 1:
      print("Premium Customer")
    else:
      print("Normal Customer")
V1 = Vehicle("001",datetime.date(2022,1,1),4)
print(V1.Calculate_age())
B1 = Bus("001",datetime.date(2022,1,1),4,40,"Dhaka-Rajshahi")
B1.Show_name_route()
C1 = Customers("001",datetime.date(2022,1,1),4,"Arafat","008","Dhaka",["001","002"])
C1.Buy_product("003")
C1.Is_premium()
```

```python
from abc import ABC, abstractmethod
class Shape(ABC):
  @abstractmethod
  def area(self):
    pass
  @abstractmethod
  def perimeter(self):
    pass
class Rectangle(Shape):
  def __init__(self,length,width):
    self.length = length
    self.width = width
  def area(self):
    return self.length*self.width
  def perimeter(self):
    return 2*(self.length+self.width)
R1 = Rectangle(10,20)
print(R1.area())
print(R1.perimeter())
```

```python
class Shape:
  def area(self, *args):
    if len(args) == 1:
      pi = 3.1416
      return pi*args[0]*args[0]
    elif len(args) == 2:
      return args[0]*args[1]
    elif len(args) == 3:
      s = sum(args)/2
      return (s*(s-args[0])*(s-args[1])*(s-args[2]))**0.5
    else:
      return None

S1 = Shape()
print("The Circle's Area is:",S1.area(10))
print("The Rectangle's Area is:",S1.area(10,20))
print("The Triangle's Area is:",S1.area(10,20,30))
```

```python
class Manager: #overriding with multiple inheritance
  def __init__(self, team_size):
    self.team_size = team_size
  def role(self):
    print(f"Managing a team of {self.team_size} members.")
class Developer:
  def __init__(self, language):
    self.language = language
  def role(self):
    print(f"Coding in {self.language} language.")
class TeamLead(Manager, Developer):
  def __init__(self, team_size, language):
    Manager.__init__(self, team_size)
    Developer.__init__(self, language)
  def role(self):
    print(f"Leading the team and coding in {self.language}")
M1 = Manager(10)
M1.role()
D1 = Developer("Python")
D1.role()
T1 = TeamLead(5,"Python")
T1.role()
```

```python
class Book:
  def __init__(self,title,author):
    self.title = title
    self.author = author
  def details(self):
    print(f"Title: {self.title}, Author: {self.author}")
class TextBook(Book):
  def __init__(self,title,author,subject):
    super().__init__(title,author)
    self.subject = subject
  def details(self):
    super().details()
    print(f"Subject: {self.subject}")
class ScienceTextBook(TextBook):
  def __init__(self,title,author,subject,editions):
    super().__init__(title,author,subject)
    self.editions = editions
  def details(self):
    super().details()
    print(f"Editions: {self.editions}")
B1 = Book("The ITM","Arafat Sakib")
B1.details()
T1 = TextBook("The ITM","Arafat Sakib","IT")
T1.details()
S1 = ScienceTextBook("The ITM","Arafat Sakib","IT",2)
S1.details()
```

```python
from multipledispatch import dispatch
class Volume:
  @dispatch(int)
  def calculate(self,l):
    return l*l*l
  @dispatch(int,int)
  def calculate(self,l,w):
    return l*w
  @dispatch(int,int,int)
  def calculate(self,l,w,h):
    return l*w*h
V1 = Volume()
print("The Cube's Volume is =",V1.calculate(10))
print("The Cuboid's Volume is =",V1.calculate(10,20))
print("The Cuboid's Volume is =",V1.calculate(10,20,30))
```

```python
class Volume:
  def calculate(self,*args):
    if len(args) == 1:
      return args[0]*args[0]*args[0]
    elif len(args) == 2:
      return args[0]*args[1]
    elif len(args) == 3:
      return args[0]*args[1]*args[2]
    else:
      return None
V1 = Volume()
print("The Cube's Volume is =",V1.calculate(10))
print("The Cuboid's Volume is =",V1.calculate(10,20))
print("The Cuboid's Volume is =",V1.calculate(10,20,30))
```

```python
class Perimeter:
  def compute(self,*args):
    if len(args) == 1:
      pi = 3.1416
      return args[0]*2*pi
    elif len(args) == 2:
      return 2*(args[0]+args[1])
    else:
      return None
P1 = Perimeter()
print("The Circle's Perimeter is =",P1.compute(10))
print("The Rectangle's Perimeter is =",P1.compute(10,20))
```

```python
from multipledispatch import dispatch
class Perimeter:
  @dispatch(int)
  def compute(self,l):
    pi = 3.1416
    return 2*pi*l
  @dispatch(int,int)
  def compute(self,l,w):
    return 2*(l+w)
P1 = Perimeter()
print("The Circle's Perimeter is =",P1.compute(10))
print("The Rectangle's Perimeter is =",P1.compute(10,20))
```
