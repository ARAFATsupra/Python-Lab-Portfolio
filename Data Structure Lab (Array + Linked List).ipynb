```python
print("\t\tWelcome To The Python Program !!\n")
N  = int(input("Enter The Number Of Array Elements: "))
A = []
print("Enter The Array Elements = ")
for i in range(N):
  element = int(input())
  A.append(element)
print("The Array Is = ",A)
Size = len(A)
print("The Size Of The Array Is = ",Size)
S = int(input("Enter The Value For Binary Search: "))
LOW = 0
HIGH = Size - 1
while LOW <= HIGH:
  MID = (LOW + HIGH) // 2
  if A[MID] == S:
    print("Element Found In The Array At Index = ",MID)
    break
  elif A[MID] < S:
    LOW = MID + 1
  else:
    HIGH = MID - 1
else:
  print("Element Not Found In The Array.")
```

```python
print("\t\tWelcome To The Python Program !!\n")
def Bubble_Sort(A):
  Size = len(A)
  for i in range(Size):
    for j in range(Size-1-i):
      if A[j] > A[j+1]:
        A[j],A[j+1] = A[j+1],A[j]
  return A
Data = [-2,45,0,11,-9]
print("The Unsorted Array Is = ",Data)
print("The Sorted Array Is = ",Bubble_Sort(Data))
```

```python
print("\t\tWelcome To The Python Program !!\n")
def Selection_Sort(A):
  Size = len(A)
  for i in range(Size):
    min_index = i
    for j in range(i+1,Size):
      if A[j] < A[min_index]:
        min_index = j
    A[i],A[min_index] = A[min_index],A[i]
  return A
Data = [-2,45,0,11,-9]
print("The Unsorted Array Is = ",Data)
print("The Sorted Array Is = ",Selection_Sort(Data))
```

```python
print("\t\tWelcome To The Python Program !!\n")
def Insertion_Sort(A):
  Size = len(A)
  for i in range(1,Size):
    key = A[i]
    j = i-1
    while j >= 0 and A[j] > key:
      A[j+1] = A[j]
      j = j-1
    A[j+1] = key
  return A
a = int(input("Enter The Number Of Array Elements: "))
Data = []
print("Enter The Array Elements = ")
for i in range(a):
  element = int(input())
  Data.append(element)
print("The Unsorted Array Is = ",Data)
Insertion_Sort(Data)
print("The Sorted Array Is = ",Data)
```
