```python
import random

print("\t\t\t\tWelcome to the Friendship Meter!!\n")

def friendship_meter(name1, name2):
    # Combine both names
    combined_names = name1.lower() + name2.lower()

    # Use the sum of the ASCII values of characters for a simple formula
    score = sum(ord(char) for char in combined_names if char.isalpha())

    # Calculate a "friendship percentage" based on the score and a random factor
    friendship_percentage = (score % 100) + random.randint(0, 10)

    # Ensure the percentage is between 0 and 100
    friendship_percentage = min(friendship_percentage, 100)

    return friendship_percentage

# Get user input
name1 = input("Enter First Person's Name: ")
name2 = input("Enter Second Person's Name: ")

# Calculate and display friendship percentage
percentage = friendship_meter(name1, name2)
print(f"\nThe Friendship Compatibility Between {name1} and {name2} is = {percentage}% !")
```

```python
class Book:
  def __init__(self, title, author, copies):
    self.__title = title
    self.__author = author
    self.__copies = copies
  def borrow_book(self):
    if self.__copies > 0:
      self.__copies -= 1
      print(f"Book '{self.__title}' borrowed successfully!")
    else:
      print(f"Book '{self.__title}' is not available for borrowing.")

  def return_book(self):
    self.__copies += 1
    print(f"Book '{self.__title}' returned successfully!")
  def get_details(self):
    return f"Title: {self.__title}\nAuthor: {self.__author}\nCopies: {self.__copies}"

B1 = Book("The Alchemist", "Paulo Coelho", 5)
print(B1.get_details())
B1.borrow_book()
print(B1.get_details())
B1.return_book()
print(B1.get_details())
print("The number of book copies is:",B1._Book__copies)
```

```python
class Product:
  def __init__(self, product_name, quantity, price_per_unit):
    self.__product_name = product_name
    self.__quantity = quantity
    self.__price_per_unit = price_per_unit
  def add_stock(self):
    self.__quantity += 1
  def sell(self):
    if self.__quantity > 0:
      self.__quantity -= 1
      print(f"Product '{self.__product_name}' sold successfully!")
    else:
      print(f"Product '{self.__product_name}' is out of stock.")
  def get_inventory_value(self):
    return self.__quantity * self.__price_per_unit

P1 = Product("Laptop", 5, 1000)
print(f"Inventory Value: ${P1.get_inventory_value()}")
P1.add_stock()
print(f"Inventory Value: ${P1.get_inventory_value()}")
P1.sell()
print("The price per unit is:",P1._Product__price_per_unit)
```

```python
from multipledispatch import dispatch
class Volume:
  @dispatch(int,int,int)
  def calculate_volume(self,length,width,height):
    return length * width * height
  @dispatch(int,int)
  def calculate_volume(self,radius,height):
    return 3.1416 * (radius ** 2) * height
  @dispatch(int)
  def calculate_volume(self,radius):
    return 4/3 * 3.1416 * (radius ** 3)
V = Volume()
print(V.calculate_volume(2,3,4))
print(V.calculate_volume(2,3))
print(V.calculate_volume(2))
```

```python
class Volume:
  def calculate_volume(self,*args):
    if len(args) == 3:
      length, width, height = args
      return length * width * height
    elif len(args) == 2:
      radius, height = args
      return 3.1416 * (radius ** 2) * height
    elif len(args) == 1:
      radius = args[0]
      return 4/3 * 3.1416 * (radius ** 3)
V = Volume()
print(V.calculate_volume(2,3,4))
print(V.calculate_volume(2,3))
print(V.calculate_volume(2))
```

```python
class Vehicle:
  def __init__(self,vehicle_name,base_rent):
    self.vehicle_name = vehicle_name
    self.base_rent = base_rent
  def calculate_rent(self,days):
    return self.base_rent * days
class Car(Vehicle):
  def __init__(self,vehicle_name,base_rent,ac_available,fuel_type):
    super().__init__(vehicle_name,base_rent)
    self.ac_available = ac_available
    self.fuel_type = fuel_type
  def calculate_rent(self,days):
    rent = super().calculate_rent(days)
    if self.ac_available:
      rent += 10
    elif self.fuel_type == "Electric":
      rent -= 5
    else:
      print("None")
class Bike(Vehicle):
  def __init__(self,vehicle_name,base_rent,gear_type,helmet_included):
    super().__init__(vehicle_name,base_rent)
    self.gear_type = gear_type
    self.helmet_included = helmet_included
  def calculate_rent(self,days):
    rent = super().calculate_rent(days)
    if self.gear_type == "Manual":
      rent += 5
    elif self.helmet_included:
      rent += 10
    else:
      print("None")
V1 = Vehicle("Toyota", 100)
print(V1.calculate_rent(5))
C1 = Car("Toyota", 100, True, "Petrol")
print(C1.calculate_rent(5))
B1 = Bike("Toyota", 100, "Manual", True)
print(B1.calculate_rent(5))
```
