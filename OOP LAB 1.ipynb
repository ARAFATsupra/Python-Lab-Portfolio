```python
class rectangle:
  def __init__(self,length,width):
    self.length = length
    self.width = width
  def area(self):
    return self.length*self.width
  def perimeter(self):
    return 2*(self.length+self.width)
rectangle1 = rectangle(5,4)
print(rectangle1.area())
print(rectangle1.perimeter())
```

```python
class employee:
  def __init__(self,name,salary):
    self.name = name
    self.salary = salary
  def display(self):
    print(f"Name: {self.name}")
    print(f"Salary: {self.salary}")
  def give_raise(self):
    self.salary = self.salary*1.02
    print(f"New salary: {self.salary}")
  def is_higher(self):
    return self.salary>80000
employee1 = employee("Arafat",50000)
employee1.display()
employee1.give_raise()
employee1.is_higher()
```

```python
def is_prime(x):
    if x < 2:
        return False
    for i in range(2, x):
        if x % i == 0:
            return False
    return True
x = int(input("Enter a number: "))
if is_prime(x):
    print(f"{x} is a prime number")
else:
    print(f"{x} is not a prime number")
```

```python
class bank_account:
  def __init__(self,name,balance):
    self.name = name
    self.balance = balance
  def deposit(self):
    amount = int(input("Enter the amount to deposit: "))
    self.balance+=amount
    print(f"New balance: {self.balance}")
  def withdraw(self):
    amount = int(input("Enter the amount to withdraw: "))
    if amount>self.balance:
      print("Insufficient balance")
    else:
      self.balance-=amount
      print(f"New balance: {self.balance}")
X = bank_account("Arafat",50000)
X.deposit()
X.withdraw()
```

```python
def categorize_temperatures():
    for temp in range(0, 101):
        if temp < 30:
            print(f"Temperature {temp}: Cold")
        elif 30 <= temp <= 60:
            print(f"Temperature {temp}: Moderate")
        else:
            print(f"Temperature {temp}: Hot")
categorize_temperatures()
```

```python
class MobilePhone:
  def __init__(self,battery,battery_percentage):
    self.battery = battery
    self.battery_percentage = battery_percentage
  def charge(self,amount):
    self.battery_percentage += amount
    print(f"New battery percentage: {self.battery_percentage}")
  def is_fully_charged(self):
    return self.battery_percentage == 100
X = MobilePhone("Samsung",5)
X.charge(5)
X.is_fully_charged()
```

```python
class Animal:
    def __init__(self, species, age):
        self.species = species
        self.age = age
    def is_adult(self):
        return self.age > 2
    def make_sound(self):
        if self.species.lower() == "lion":
            print("Roar")
        elif self.species.lower() == "dog":
            print("Bark")
        elif self.species.lower() == "cat":
            print("Meow")
        else:
            print("Some generic animal sound")
animal1 = Animal("Lion", 3)
animal2 = Animal("Dog", 1)
print(f"Is {animal1.species} an adult? {animal1.is_adult()}")
animal1.make_sound()
print(f"Is {animal2.species} an adult? {animal2.is_adult()}")
animal2.make_sound()
```
