```python
A = int(input("Enter the number of students: "))
S = []
for i in range(A):
  Element = int(input("Enter the scores: "))
  S.append(Element)
print("The scores are: ",S)
def Bubble_Sort(A):
  for i in range(len(A)-1):
    for j in range(len(A)-1):
      if A[j] > A[j+1]:
        A[j],A[j+1] = A[j+1],A[j]
  return A
print("The Sorted Scores Are: ",Bubble_Sort(S))
```

```python
A = int(input("Enter the number of students: "))
S = []
for i in range(A):
  Element = int(input("Enter the roll number: "))
  S.append(Element)
print("The roll numbers are: ",S)
def Bubble_Sort(A):
  for i in range(len(A)-1):
    for j in range(len(A)-1):
      if A[j] > A[j+1]:
        A[j],A[j+1] = A[j+1],A[j]
  return A
print("The Sorted Roll Numbers Are: ",Bubble_Sort(S))
B = int(input("Enter the roll number to search by binary search: "))
def Binary_Search(A,B):
  low = 0
  high = len(A)-1
  while low <= high:
    mid = (low+high)//2
    if A[mid] == B:
      return mid
    elif A[mid] < B:
      low = mid+1
    else:
      high = mid-1
  return -1
result = Binary_Search(S,B)
if result != -1:
  print(f"The roll number {B} is found at index",result)
else:
  print(f"The roll number {B} is not found in the list")
```

```python
class Node:
  def __init__(self,data):
    self.data = data
    self.next = None
n = int(input("Enter the number of courses: "))
head = None
for i in range(n):
  course = int(input(f"Enter the course code {i}: "))
  if head is None:
    head = Node(course)
  else:
    A = head
    while A.next:
      A = A.next
    A.next = Node(course)
    A = A.next
print("The list of course codes: ")
A = head
while A is not None:
  print(A.data,end = " ")
  A = A.next
print()
B = int(input("Enter the new course code to insert at given position: "))
C = int(input("Enter the position to insert the new course code: "))
A = head
for i in range(C-2):
  A = A.next
new_node = Node(B)
new_node.next = A.next
A.next = new_node
print("The updated list of course codes: ")
A = head
while A is not None:
  print(A.data,end = " -> ")
  A = A.next
print()
```
