```python
# Initial string
P = "ppp ppp ppp newyork ppp ppp is my ppp home town ppp"

# Replace "newyork" with "Barisal" (or your hometown)
P = P.replace("newyork", "Barisal")

# Remove "ppp" from the string
P = P.replace("ppp", "")

# Split the string by spaces into a list
words = P.split()

# Rebuild the string using a for loop and avoid adding extra spaces at the end
final_string = ""
for i in range(len(words)):
    if i == len(words) - 1:  # For the last word, don't add an extra space
        final_string += words[i]
    else:
        final_string += words[i] + " "

# Print the final string
print(final_string)
```

```python
#Constructor Example
class Room:
  def __init__(self, length, breadth):
    self.length = length
    self.breadth = breadth
  def calculate_area(self):
    print("Area of Room =", self.length * self.breadth)
  def Display(self):
    print("Length: ", self.length)
    print("Breadth: ", self.breadth)
study_room = Room(42.5, 30.8)
study_room.calculate_area()
Bedroom = Room(42.5, 30.8)
Bedroom.calculate_area()
Classroom = Room(60, 40)
Classroom.calculate_area()
study_room.Display()
Bedroom.Display()
Classroom.Display()
```

```python
class Laptop:
    def __init__(self, brand, battery_percentage):
        self.brand = brand
        self.battery_percentage = battery_percentage
    def needs_charging(self):
        return self.battery_percentage < 20
    def use_battery(self, amount):
        self.battery_percentage -= amount
#Object Create
laptop1 = Laptop("Dell", 50)
laptop2 = Laptop("HP", 20)
print(laptop1.needs_charging())
laptop2.use_battery(5)
print(laptop2.battery_percentage)
print(laptop2.needs_charging())
```

```python
class ShoppingCart:
  def __init__(self):
    self.items = []
  def add_item(self, item):
    self.items.append(item)
  def remove_item(self, item):
    self.items.remove(item)
  def total_items(self):
    return len(self.items)
Cart1 = ShoppingCart()
Cart1.add_item("Shirt")
Cart1.add_item("Pant")
Cart1.add_item("Shoes")
print(f"Total items in Cart1: {Cart1.total_items()}")
Cart1.remove_item("Pant")
print(f"Total items in Cart1: {Cart1.total_items()}")
```

```python
class Book:
  def __init__(self,title,author,pages):
    self.title = title
    self.author = author
    self.pages = pages
  def is_long(self):
    if self.pages > 300:
      return True
    else:
      return False
  def Display(self):
    print(f"Title: {self.title}, Author: {self.author}, Pages: {self.pages}")
Book1 = Book("ITM", "Arafat", 500)
print(Book1.is_long())
Book1.Display()
Book2 = Book("ITM", "Arafat", 100)
print(Book2.is_long())
Book2.Display()
```

```python
class Account:
  def __init__(self,Account_number,Current_balance,Current_loan):
    self.Account_number = Account_number
    self.Current_balance = Current_balance
    self.Current_loan = Current_loan
  def Display(self):
    print(f"Account Number: {self.Account_number}, Current Balance: {self.Current_balance}, Current Loan: {self.Current_loan}")
  def Calculate_interest(self):
    P = self.Current_balance
    r = float(input("Enter interest rate: "))
    n = 1
    I = P*r*n
    self.Current_balance = P + I
    print(f"New Balance: {self.Current_balance}")
    print(f"Interest: {I}")
  def is_rich(self):
    if self.Current_balance > self.Current_loan:
      print("Yes")
    else:
      print("No")
Account1 = Account(123456789, 10000, 5000)
Account1.Display()
print(f"Interest: {Account1.Calculate_interest()}")
Account1.Display()
Account1.is_rich()
```

```python
class Car:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed
    def is_moving(self):
        return self.speed > 0
    def accelerate(self, amount):
        self.speed += amount
#Object Create
car1 = Car("Toyota", 50)
car2 = Car("Honda", 0)
print(car1.is_moving())
car2.accelerate(20)
print(car2.speed)
```
